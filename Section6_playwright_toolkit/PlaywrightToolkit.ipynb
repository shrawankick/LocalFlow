```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    base_url="http://localhost:11434",
    model="gpt-oss:20b-cloud",
    temperature=0,
    timeout=300,
    num_ctx=8192,
    keep_alive="1h"
).bind_tools(tools)
```

```python
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage
#print(f"Tools available to the agent: {tools}")
agent = create_agent(
    tools = tools,
    model = llm
)
query = "Go to https://www.w3schools.com/ So I want you to go through the web page and find out all the tutorials (whatever they are). Also identify what are the top tutorials that they are using. After identifying the top tutorials please close the browser. And give me everything in a text format - link. Also top tutorials name and link. Close the browser after providing the response. in abluar format only text."
##query = "Go to https://example.com and tell me what the page title is."
##query = "What is the result of adding 5 and 3, then multiplying the result by 2, and finally subtracting 4?"
response = await agent.ainvoke({"messages": [HumanMessage(content=query)]})
print("Query:", query)
print("Final Result:", response)
```

Step-by-Step Explanation
1. Imports and Prerequisites
from langchain.agents import create_agent: Imports the factory function used to build the agent logic.
from langchain_core.messages import HumanMessage: Imports a class used to structure the user's input in a chat-friendly format.
Prerequisites (llm and tools): The code relies on two external variables defined earlier in your script:
llm: An instance of ChatOllama configured to use structured JSON output (e.g., format="json" and .bind_tools()). This is the AI's "brain."
tools: A list of functions/classes (like browser navigation, text extraction, etc.) that the AI can choose to execute. This is the AI's "hands."
2. Agent Initialization
python
agent = create_agent(
    tools = tools,
    model = llm
)
Use code with caution.

This section instantiates the agent. It tells the agent framework which model to use for reasoning (llm) and which capabilities it possesses (tools). The agent uses this configuration to decide its next action in an autonomous loop (Think -> Act -> Observe).
3. Defining the Goal/Query
python
query = "Go to www.w3schools.com So I want you to go through the web page and find out all the tutorials..."
Use code with caution.

This variable holds the detailed instruction for the agent. The clarity and detail of this prompt are crucial for complex agent tasks like web browsing.
4. Invoking the Agent (The Execution Phase)
python
response = await agent.ainvoke({"messages": [HumanMessage(content=query)]})
Use code with caution.

This is where the execution happens:
await: This keyword is essential because the agent process is asynchronous (it takes time to run the browser, make decisions, etc.). await ensures the Python script pauses and waits for the entire process to complete before moving to the print statements.
agent.ainvoke(...): Starts the agent's autonomous reasoning and action loop based on the provided query. The input format {"messages": [...]} is required by this specific agent configuration.
5. Printing the Result (Debugging Output)
python
print("Final Result:", response)
Use code with caution.

This line prints the entire raw data structure (a Python dictionary) that the agent framework returns upon completion.
Note: This is a debugging print statement. In production code, you typically extract just the final text answer using print("Final Result:", response["messages"][-1].content).
