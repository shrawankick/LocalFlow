### Working with Playwright 

So when you are installing the Playwright in the new machine, you might get this error. At the end you might get the error that the browser is not available or something like that. So when you are doing that what you need to do 
ince you see .venv in your PowerShell prompt (e.g., (.venv) PS C:\Users\YourName\YourProject>), you are already inside your virtual environment. You do not need to go anywhere else.

Here are the specific instructions to fix your installation right now.

Step 1: Run these two commands
Type these commands into your PowerShell terminal one by one, pressing Enter after each line.

Command 1: Install the Python Library This downloads the Playwright code so Python can talk to browsers.

PowerShell

`pip install playwright`
Command 2: Install the Browser Engines This is the step most people forget. It downloads the actual browser binaries (Chromium, Firefox, WebKit) that the code will control.

PowerShell

`playwright install`
(Note: This second command might take a minute or two as it downloads about 100-200MB of browser files.)

Step 2: Restart Your Kernel
After the installation finishes in PowerShell, go back to your code editor (VS Code/Jupyter):

Look at the top right of your notebook.

Click the Restart button (usually a circular arrow icon ↻).

This forces your notebook to "reload" and see the new libraries you just installed.

Step 3: Verify with this Simple Test
Once you have restarted, run this code in a new cell. If it prints the success message, you are 100% ready to go.

Python
`
from playwright.async_api import async_playwright

print("Playwright module is successfully installed!")`
Why did you need to do this?
Think of it like buying a remote control car:

pip install playwright is buying the Remote Control (the Python code).

playwright install is buying the Car itself (the Chromium browser).

You need both to make it drive!

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    base_url="http://localhost:11434",
    model="gpt-oss:20b-cloud",
    temperature=0,
    timeout=300,
    num_ctx=8192,
    keep_alive="1h"
).bind_tools(tools)
```

```python
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage

print(f"Tools available to the agent: {tools}")


agent = create_agent(
    tools = tools,
    model = llm
)

query = "Go to https://www.w3schools.com/ So I want you to go through the web page and find out all the tutorials (whatever they are). Also identify what are the top tutorials that they are using. After identifying the top tutorials please close the browser. And give me everything in a text format - link. Also top tutorials name and link. Close the browser after providing the response. in abluar format only text."
##query = "Go to https://example.com and tell me what the page title is."

##query = "What is the result of adding 5 and 3, then multiplying the result by 2, and finally subtracting 4?"
response = await agent.ainvoke({"messages": [HumanMessage(content=query)]})
print("Query:", query)
#print("Final Result:", response['output'])
print("Final Result:", response)
#result =  await agent.ainvoke({"messages": [HumanMessage(content=query)]})
##result =  await agent.invoke({"input": query})
##print("Query:", query)
##print("Final Result:", result['output'])
#print("Final Result:", result["messages"][-1].content)
#print(result['output'])
```
